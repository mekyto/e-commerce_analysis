```python
import pandas as pd
import numpy as np
```

```python
df_item_refunds=pd.read_csv('/Users/keneskenesbay/Documents/project_1/Maven+Fuzzy+Factory/order_item_refunds.csv')
df_order_items=pd.read_csv('/Users/keneskenesbay/Documents/project_1/Maven+Fuzzy+Factory/order_items.csv')
df_order=pd.read_csv('/Users/keneskenesbay/Documents/project_1/Maven+Fuzzy+Factory/orders.csv')
df_products=pd.read_csv('/Users/keneskenesbay/Documents/project_1/Maven+Fuzzy+Factory/products.csv')
df_websitepage_view=pd.read_csv('/Users/keneskenesbay/Documents/project_1/Maven+Fuzzy+Factory/website_pageviews.csv')
df_website_sessions=pd.read_csv('/Users/keneskenesbay/Documents/project_1/Maven+Fuzzy+Factory/website_sessions.csv')
```

```python
#Checking sum of NA values in the dataframes
display("df item refunds",df_item_refunds.isna().sum(),
       "df order items",df_order_items.isna().sum(),
       "df order",df_order.isna().sum(),
       "df products",df_products.isna().sum(),
       "df websitepage view ",df_websitepage_view.isna().sum(),
       "df website sessions",df_website_sessions.isna().sum())
```

```python
#Checking sum of the duplicated values in the dataFrames
display("df item refunds",df_item_refunds.duplicated().sum(),
       "df order items",df_order_items.duplicated().sum(),
       "df order",df_order.duplicated().sum(),
       "df products",df_products.duplicated().sum(),
       "df websitepage view ",df_websitepage_view.duplicated().sum(),
       "df website sessions",df_website_sessions.duplicated().sum())
```

```python
#Checking sum of the duplicate values in primary key of the dataFrames
display(df_item_refunds["order_item_refund_id"].duplicated().sum(),
df_order_items["order_item_id"].duplicated().sum(),
df_order["order_id"].duplicated().sum(),
df_products["product_id"].duplicated().sum(),
df_websitepage_view["website_pageview_id"].duplicated().sum(),
df_website_sessions["website_session_id"].duplicated().sum())
```

```python
#checking the data types of dataFrames
display(df_item_refunds.info(),
        df_order_items.info(),
        df_order.info(),
        df_products.info(),
        df_websitepage_view.info(),
        df_website_sessions.info())
```

```python
#Change types of dataframes to the correct 
df_item_refunds['created_at']=pd.to_datetime(df_item_refunds['created_at'])
df_order_items['created_at']=pd.to_datetime(df_order_items['created_at'])
df_order['created_at']=pd.to_datetime(df_order['created_at'])
df_products['created_at']=pd.to_datetime(df_products['created_at'])
df_websitepage_view['created_at']=pd.to_datetime(df_websitepage_view['created_at'])
df_website_sessions['created_at']=pd.to_datetime(df_website_sessions['created_at'])
```

```python
#Checking are there any negative numbers in the order prices?
display(
    (df_order['price_usd']<0).sum(),
    (df_order['cogs_usd']<0).sum()
)
```
